```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
model = SentenceTransformer('sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2')

# Извлечение наименований программ
program_names = programs_df.tolist()
```

```python
data = program_names
programs_df = pd.DataFrame(data)
```

```python
embeddings = model.encode(program_names, convert_to_tensor=True)
embeddings_np = embeddings.numpy()
```

```python
def find_closest_name(query: str, top_k: int = 1) -> pd.DataFrame:
    """
    Находит top_k ближайших названий из programs_df для заданного запроса.
    Возвращает DataFrame с результатами.
    """
    # Кодируем запрос
    query_embedding = model.encode(query, convert_to_tensor=True)
    query_embedding_np = query_embedding.numpy().reshape(1, -1)

    # Считаем косинусную схожесть
    similarities = cosine_similarity(query_embedding_np, embeddings_np)[0]

    # Сортируем по убыванию схожести
    sorted_indices = np.argsort(similarities)[::-1]

    # Формируем результат
    results = []
    for idx in sorted_indices[:top_k]:
        results.append({
            'Запрос': query,
            'Ближайшее название': program_names[idx],
            'Схожесть': f"{similarities[idx]:.4f}"
        })

    return pd.DataFrame(results)
```

```python
query = "Приклыадная информатика в информациавонной сфере"
result_df = find_closest_name(query, top_k=3)
print(result_df)
```
